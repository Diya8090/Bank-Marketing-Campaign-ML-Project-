### ✅ Task 8 — Adding a New Column
Add a new column called **`total_revenue`** computed as:
> `total_revenue = quantity × unit_price × (1 - discount / 100)`

> ⚠️ Treat missing `discount` values as **0** (no discount).

Print the first 10 rows showing `product`, `quantity`, `unit_price`, `discount`, and `total_revenue`.

```python
df['total_revenue'] = (df['quantity'] * df['unit_price'] * (1-df['discount'].fillna(0)/100))
df[['product','quantity','discount','total_revenue']].head(10)
```
