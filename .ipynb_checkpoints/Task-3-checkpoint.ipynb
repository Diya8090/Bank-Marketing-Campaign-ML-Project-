# Linear Regression on Car Data set

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("MacBook_Air_Dataset.csv")
df
```

```python
X = df[['number']]
y = df['Price']
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
```

```python
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
prediction = model.predict([[8]])

print("Predicted Price:", prediction[0])
```

```python
predicted = model.predict(X)
predicted
```

```python
plt.scatter(X,y)

plt.xlabel("Model Number")

plt.ylabel("Price")

plt.title("MacBook Air Dataset")

plt.show()
```

```python
plt.scatter(X,y,color='blue')

plt.plot(X,model.predict(X),color='red')

plt.xlabel("Model Number")

plt.ylabel("Price")

plt.title("Linear Regression")

plt.show()
```

## PART B – Gradient Descent From Scratch

```python
X = df['number'].values.astype(float)

Y = df['Price'].values.astype(float)

n = len(X)
```

```python
m = 0

c = 0

lr = 0.01

epochs = 10000

loss_history = []
```

```python
for i in range(epochs):

    Y_pred = m * X + c

    error = Y_pred - Y

    loss = (1/n) * np.sum(error ** 2)

    loss_history.append(loss)

    dm = (2/n) * np.sum(error * X)

    dc = (2/n) * np.sum(error)

    m = m - lr * dm

    c = c - lr * dc

    if i % 1000 == 0:
        print("Epoch:", i, "Loss:", loss)
```

```python
print("Slope =",m)

print("Intercept =",c)
```

```python
new_model = 8

predicted_price = m * new_model + c

print(predicted_price)
```

```python
plt.plot(loss_history)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title("Loss vs Epoch")

plt.show()
```

```python
print("Sklearn Slope =",model.coef_[0])

print("Gradient Descent Slope =",m)

print()

print("Sklearn Intercept =",model.intercept_)

print("Gradient Descent Intercept =",c)
```
