### Converting column's datatype to numeic

```python
num_cols = [
    'balance',
    'day',
    'duration',
    'campaign',
    'pdays',
    'previous'
]
```

```python
for col in num_cols:
    df[col] = pd.to_numeric(df[col])

df.info()
```

### Detect Outliers

```python
for col in num_cols:

    plt.figure(figsize=(6,3))

    sns.boxplot(x=df[col])

    plt.title(f"Boxplot of {col}")

    plt.show()
```

### Handle Outliers

```python
for col in num_cols:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR

    upper = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower) & (df[col] <= upper)]

print(df.shape)
```

### Correlation Heatmap

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

cat_cols = [
    'job',
    'marital',
    'education',
    'default',
    'housing',
    'loan',
    'contact',
    'month',
    'poutcome',
    'y'
]

for col in cat_cols:
    df[col] = le.fit_transform(df[col])
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(df.corr(), annot=True, cmap="coolwarm")

plt.title("Correlation Heatmap")

plt.show()
```
